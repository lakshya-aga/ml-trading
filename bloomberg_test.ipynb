```python
import blpapi
from datetime import datetime


TICKER = "NIFTY Index"

HOST = "localhost"
PORT = 8194


def main():

    # Connect to Bloomberg Terminal
    session_options = blpapi.SessionOptions()
    session_options.setServerHost(HOST)
    session_options.setServerPort(PORT)

    session = blpapi.Session(session_options)

    if not session.start():
        raise RuntimeError("Could not start Bloomberg session")

    print("Connected to Bloomberg")

    if not session.openService("//blp/refdata"):
        raise RuntimeError("Could not open //blp/refdata")

    service = session.getService("//blp/refdata")

    # Historical tick request
    request = service.createRequest("IntradayTickRequest")

    request.set("security", TICKER)

    # Date/time
    request.set(
        "startDateTime",
        datetime(2025, 1, 2, 9, 15)
    )

    request.set(
        "endDateTime",
        datetime(2025, 1, 2, 15, 30)
    )

    # Ask for trades
    request.getElement("eventTypes").appendValue("TRADE")

    # Include condition codes
    request.set("includeConditionCodes", True)

    print(request)

    session.sendRequest(request)

    while True:

        event = session.nextEvent()

        for msg in event:

            if msg.hasElement("responseError"):
                print(msg)
                continue

            if msg.messageType() == blpapi.Name("IntradayTickResponse"):

                tick_data = msg.getElement("tickData")
                ticks = tick_data.getElement("tickData")

                for i in range(ticks.numValues()):

                    tick = ticks.getValueAsElement(i)

                    print(tick)

        if event.eventType() == blpapi.Event.RESPONSE:
            break

    session.stop()


if __name__ == "__main__":
    main()
```
